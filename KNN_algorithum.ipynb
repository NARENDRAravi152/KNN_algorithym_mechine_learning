# import library

```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split,KFold,cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score

import warnings
warnings.filterwarnings("ignore")
```

# import data

```python
wine_data = load_wine()
wine_data
```

```python
wine_data_df = pd.DataFrame(wine_data.data,columns=wine_data.feature_names)
wine_data_df["Target"] = wine_data.target
wine_data_df
```

```python
X = wine_data_df.drop("Target",axis=1)
y = wine_data_df.Target
```

# data preparations

## we must do an standard or normalized in input features

```python
std_scalar = StandardScaler()
scaled_X = pd.DataFrame(std_scalar.fit_transform(X))
X_train,X_test,y_train,y_test=train_test_split(scaled_X,y,train_size=0.80,shuffle=True,random_state=129,stratify=y)
```

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

# model buildiing

```python
knn_classifiers = KNeighborsClassifier(n_neighbors=29)
knn_classifiers
knn_classifiers.fit(X_train,y_train)
Y_pred_train_knn = knn_classifiers.predict(X_train)
print("confusion metrix:\n",confusion_matrix(y_train,Y_pred_train_knn))
print("accuracy score:",round(accuracy_score(y_train,Y_pred_train_knn),4))
print("precision score:",round(precision_score(y_train,Y_pred_train_knn,average="weighted"),4))
print("recall score:",round(recall_score(y_train,Y_pred_train_knn,average="weighted"),4))
```

# test data

```python
knn_classifiers = KNeighborsClassifier(n_neighbors=29)
knn_classifiers
knn_classifiers.fit(X_test,y_test)
Y_pred_test_knn = knn_classifiers.predict(X_test)
print("confusion metrix:\n",confusion_matrix(y_test,Y_pred_test_knn))
print("accuracy score:",round(accuracy_score(y_test,Y_pred_test_knn),4))
print("precision score:",round(precision_score(y_test,Y_pred_test_knn,average="weighted"),4))
print("recall score:",round(recall_score(y_test,Y_pred_test_knn,average="weighted"),4))
```

# Must to choose optimal n_neighbors

## we can use to find  for loop basis
1. for loop range(3,60)
2. knn_neighbours(i)
3. cross_val_score
4. mean ()
5.stored in list[]

```python
k_fold = KFold(n_splits=5, shuffle=True, random_state=172)
cv_scores_mean= []
for i in range(3,61,2):
    knn_classifiers = KNeighborsClassifier(n_neighbors=i)
    knn_classifiers.fit(scaled_X,y)
    cv_score  = cross_val_score(estimator=knn_classifiers,X=scaled_X,y=y,cv=k_fold)
    cv_scores_mean.append(round(cv_score.mean(),2))
    max(cv_scores_mean)
    cv_scores_mean.index(max(cv_scores_mean))
```
